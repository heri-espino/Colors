# Histograms and distributions

Histograms, smoothed densities, ECDFs, and cumulative views.

```python
import numpy as np
import matplotlib.pyplot as plt
import contrastcolors as cc

cc.set_style("heri", font="DejaVu Sans")
rng=np.random.default_rng(8)
samples=[rng.normal(-0.6,0.72,700),rng.normal(0.45,0.58,700),rng.normal(1.25,0.48,700)]
colors=cc.color_palette(hues=[20,145,210],ratio=1.2,start_luminance=0.70)

fig,ax=plt.subplots(figsize=(8.2,4.1))
for i,(sample,color) in enumerate(zip(samples,colors)):
    ax.hist(sample,bins=32,density=True,alpha=0.48,color=color,label=f"Group {i+1}")
ax.set(title="Overlaid histograms",xlabel="Value",ylabel="Density")
ax.legend()
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
def smooth_density(values,bins=120,sigma_bins=3.0):
    counts,edges=np.histogram(values,bins=bins,density=True)
    centers=0.5*(edges[:-1]+edges[1:])
    radius=int(4*sigma_bins)
    xk=np.arange(-radius,radius+1)
    kernel=np.exp(-0.5*(xk/sigma_bins)**2)
    kernel/=kernel.sum()
    return centers,np.convolve(counts,kernel,mode="same")

fig,ax=plt.subplots(figsize=(8.2,4.1))
for i,(sample,color) in enumerate(zip(samples,colors)):
    xx,yy=smooth_density(sample)
    ax.plot(xx,yy,color=color,linewidth=2,label=f"Group {i+1}")
    ax.fill_between(xx,0,yy,color=color,alpha=0.12)
ax.set(title="Smoothed density curves",xlabel="Value",ylabel="Density")
ax.legend()
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(8.2,4.1))
for i,(sample,color) in enumerate(zip(samples,colors)):
    x=np.sort(sample)
    y=np.arange(1,len(x)+1)/len(x)
    ax.step(x,y,where="post",color=color,label=f"Group {i+1}")
ax.set(title="Empirical CDF",xlabel="Value",ylabel="Probability")
ax.legend()
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(8.2,4.1))
ax.hist(samples[1],bins=35,cumulative=True,density=True,histtype="stepfilled",alpha=0.35,color=colors[1])
ax.hist(samples[1],bins=35,cumulative=True,density=True,histtype="step",linewidth=2,color=colors[1])
ax.set(title="Cumulative histogram",xlabel="Value",ylabel="Cumulative share")
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```
