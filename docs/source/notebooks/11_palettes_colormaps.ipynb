# Palettes and colormaps

Palette sizes, hue order, ContrastGrid, the Heri palette, and continuous colormaps.

```python
import numpy as np
import matplotlib.pyplot as plt
import contrastcolors as cc

cc.set_style("heri", font="DejaVu Sans")
configs=[
    ([35,155,260],1.25,"3 colors"),
    ([55,20,145,210,290],1.18,"5 colors"),
    ([20,70,120,170,220,275,330],1.10,"7 colors"),
]
fig,axes=plt.subplots(3,1,figsize=(8.4,3.2))
for ax,(hues,ratio,title) in zip(axes,configs):
    colors=cc.color_palette(hues=hues,ratio=ratio,start_luminance=0.72)
    for i,color in enumerate(colors):
        ax.barh(0,1,left=i,color=color,height=1)
    ax.set_xlim(0,len(colors)); ax.set_ylim(-0.5,0.5)
    ax.set_title(title,loc="left"); ax.axis("off")
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
grid=cc.contrast_grid(levels=5,hues=[55,20,145,210,290],ratio=1.18,start_luminance=0.72)
ax=grid.plot()
ax.set_title("ContrastGrid: luminance rows x hue columns")
ax.figure.tight_layout()
```

```python
cc.show_accessibility_panel(ax.figure, max_width=640)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(8.4,1.7))
for i,color in enumerate(cc.HERI_PALETTE):
    ax.barh(0,1,left=i,color=color,height=1)
    ax.text(i+0.5,0,color,ha="center",va="center",color="white" if i<2 else "#222222",fontsize=8)
ax.set_xlim(0,len(cc.HERI_PALETTE)); ax.set_ylim(-0.5,0.5); ax.axis("off")
ax.set_title("HERI_PALETTE",loc="left")
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
gradient=np.linspace(0,1,512)[None,:]
fig,ax=plt.subplots(figsize=(8.4,1.8))
ax.imshow(gradient,aspect="auto",cmap=cc.HERI_CMAP)
ax.set_yticks([]); ax.set_xticks([0,128,256,384,511],["0",".25",".5",".75","1"])
ax.set_title("HERI_CMAP",loc="left")
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```
