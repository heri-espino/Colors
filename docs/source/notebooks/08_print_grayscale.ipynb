# Grayscale and print-safe design

Why redundant color, luminance, marker, and line-style encodings matter.

```python
import numpy as np
import matplotlib.pyplot as plt
import contrastcolors as cc

cc.set_style("heri", font="DejaVu Sans")
x=np.linspace(0,10,280)
scheme=cc.plot_scheme(
    hues=[55,20,145,210,290],ratio=1.18,start_luminance=0.72,
    markers=["o","s","^","D","X"],
    linestyles=["-","--",":","-.",(0,(5,2))]
)
fig,ax=plt.subplots(figsize=(8.4,4.2))
for i,kwargs in enumerate(scheme):
    ax.plot(x,np.sin(x*0.7+i*0.45)+i*0.36,markevery=50,label=f"Series {i+1}",**kwargs)
ax.set(title="Color view",xlabel="x",ylabel="Response")
ax.legend(ncol=3)
fig.tight_layout()
```

```python
palette=cc.contrast_palette(hues=[55,20,145,210,290],ratio=1.18,start_luminance=0.72)

def gray_from_luminance(y):
    if y<=0.0031308:
        g=12.92*y
    else:
        g=1.055*(y**(1/2.4))-0.055
    return (g,g,g,1.0)

gray_colors=[gray_from_luminance(y) for y in palette.luminances]
fig,ax=plt.subplots(figsize=(8.4,4.2))
for i,(kwargs,gray) in enumerate(zip(scheme,gray_colors)):
    local=dict(kwargs); local["color"]=gray
    ax.plot(x,np.sin(x*0.7+i*0.45)+i*0.36,markevery=50,label=f"Series {i+1}",**local)
ax.set(title="Ideal luminance-preserving grayscale",xlabel="x",ylabel="Response")
ax.legend(ncol=3)
fig.tight_layout()
```

```python
fig,axes=plt.subplots(1,2,figsize=(10.0,3.8))
for i,kwargs in enumerate(scheme):
    y=np.sin(x*0.7+i*0.45)+i*0.36
    axes[0].plot(x,y,markevery=55,**kwargs)
for i,(kwargs,gray) in enumerate(zip(scheme,gray_colors)):
    y=np.sin(x*0.7+i*0.45)+i*0.36
    local=dict(kwargs); local["color"]=gray
    axes[1].plot(x,y,markevery=55,**local)
axes[0].set_title("Color"); axes[1].set_title("Grayscale")
for ax in axes:
    ax.set_xlabel("x"); ax.set_ylabel("Response")
fig.tight_layout()
```
