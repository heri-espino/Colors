# Lines and time series

Line plots, redundant identifiers, rolling trends, and step series.

```python
import numpy as np
import matplotlib.pyplot as plt
import contrastcolors as cc

cc.set_style("heri", font="DejaVu Sans")
x=np.linspace(0,12,360)
scheme=cc.plot_scheme(
    hues=[55,20,145,210,290], ratio=1.18, start_luminance=0.72,
    markers=["o","s","^","D","X"],
    linestyles=["-","--",":","-.",(0,(5,2))]
)
fig,ax=plt.subplots(figsize=(8.4,4.2))
for i,kwargs in enumerate(scheme):
    y=np.sin(0.62*x+i*0.52)+i*0.36
    ax.plot(x,y,markevery=55,label=f"Series {i+1}",**kwargs)
ax.set(title="Multiple line series",xlabel="Time",ylabel="Response")
ax.legend(ncol=3)
fig.tight_layout()
```

```python
rng=np.random.default_rng(4)
t=np.arange(120)
raw=np.cumsum(rng.normal(0,0.17,len(t)))+0.012*t
window=12
smooth=np.convolve(raw,np.ones(window)/window,mode="valid")

fig,ax=plt.subplots(figsize=(8.4,4.0))
ax.plot(t,raw,color=cc.HERI_NEUTRAL["mid"],alpha=0.65,label="Raw")
ax.plot(t[window-1:],smooth,color=cc.HERI_PALETTE[1],linewidth=2.2,label="Rolling mean")
ax.set(title="Time series + smooth trend",xlabel="Period",ylabel="Level")
ax.legend()
fig.tight_layout()
```

```python
events=np.array([0,1,1,2,2,3,2,4,4,5,3,3,5,6])
fig,ax=plt.subplots(figsize=(8.4,3.5))
ax.step(np.arange(len(events)),events,where="mid",color=cc.HERI_PALETTE[2],linewidth=2)
ax.scatter(np.arange(len(events)),events,color=cc.HERI_PALETTE[0],s=26,zorder=3)
ax.set(title="Step / event series",xlabel="Index",ylabel="State")
fig.tight_layout()
```
