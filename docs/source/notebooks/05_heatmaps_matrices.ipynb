# Heatmaps, matrices, and contours

Continuous scalar fields, correlation matrices, contours, and pcolormesh.

```python
import numpy as np
import matplotlib.pyplot as plt
import contrastcolors as cc

cc.set_style("heri", font="DejaVu Sans")
a=np.linspace(-2.4,2.4,90); b=np.linspace(-1.8,1.8,66)
X,Y=np.meshgrid(a,b)
Z=np.exp(-0.35*(X**2+Y**2))*np.cos(2.5*X)+0.14*Y
fig,ax=plt.subplots(figsize=(8.2,4.3))
im=ax.imshow(Z,origin="lower",aspect="auto",cmap=cc.HERI_CMAP,extent=[a.min(),a.max(),b.min(),b.max()])
fig.colorbar(im,ax=ax,label="Value")
ax.set(title="Heatmap",xlabel="x",ylabel="y")
fig.tight_layout()
```

```python
rng=np.random.default_rng(5)
A=rng.normal(size=(7,260))
A[1]=0.7*A[0]+0.4*rng.normal(size=260)
A[4]=-0.5*A[2]+0.5*rng.normal(size=260)
corr=np.corrcoef(A)
fig,ax=plt.subplots(figsize=(6.2,5.2))
im=ax.imshow(corr,vmin=-1,vmax=1,cmap="coolwarm")
ax.set_xticks(range(7),[f"V{i+1}" for i in range(7)])
ax.set_yticks(range(7),[f"V{i+1}" for i in range(7)])
fig.colorbar(im,ax=ax,label="Correlation")
ax.set_title("Correlation matrix")
fig.tight_layout()
```

```python
fig,ax=plt.subplots(figsize=(8.2,4.3))
levels=np.linspace(Z.min(),Z.max(),16)
cf=ax.contourf(X,Y,Z,levels=levels,cmap=cc.HERI_CMAP)
ax.contour(X,Y,Z,levels=levels[::2],colors="#444444",linewidths=0.45,alpha=0.65)
fig.colorbar(cf,ax=ax,label="Value")
ax.set(title="Filled contours",xlabel="x",ylabel="y")
fig.tight_layout()
```

```python
x=np.linspace(0,6,90); y=np.linspace(0,4,65)
X2,Y2=np.meshgrid(x,y); Z2=np.sin(X2)*np.cos(1.3*Y2)
fig,ax=plt.subplots(figsize=(8.2,4.3))
mesh=ax.pcolormesh(X2,Y2,Z2,shading="auto",cmap=cc.HERI_CMAP)
fig.colorbar(mesh,ax=ax,label="Value")
ax.set(title="Pcolormesh",xlabel="x",ylabel="y")
fig.tight_layout()
```
