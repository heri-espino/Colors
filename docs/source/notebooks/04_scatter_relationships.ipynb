# Scatter, regression, and density relationships

Grouped scatter plots, fitted lines, bubble plots, and hexbin density.

```python
import numpy as np
import matplotlib.pyplot as plt
import contrastcolors as cc

cc.set_style("heri", font="DejaVu Sans")
rng=np.random.default_rng(21)
colors=cc.color_palette(hues=[20,145,210],ratio=1.18,start_luminance=0.68)
fig,ax=plt.subplots(figsize=(8.2,4.3))
for i,color in enumerate(colors):
    x=rng.normal(i*0.8,0.8,100)
    y=0.55*x+rng.normal(i*0.35,0.55,100)
    ax.scatter(x,y,color=color,alpha=0.68,s=25,label=f"Group {i+1}")
ax.set(title="Grouped scatter",xlabel="x",ylabel="y")
ax.legend()
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
x=rng.normal(size=180)
y=0.75*x+0.22*x**2+rng.normal(scale=0.55,size=len(x))
coef1=np.polyfit(x,y,1); coef2=np.polyfit(x,y,2)
xx=np.linspace(x.min(),x.max(),200)

fig,ax=plt.subplots(figsize=(8.2,4.3))
ax.scatter(x,y,color=cc.HERI_NEUTRAL["mid"],alpha=0.45,s=20)
ax.plot(xx,np.polyval(coef1,xx),color=cc.HERI_PALETTE[1],label="Linear fit")
ax.plot(xx,np.polyval(coef2,xx),color=cc.HERI_PALETTE[0],linestyle="--",label="Quadratic fit")
ax.set(title="Scatter + fitted trends",xlabel="x",ylabel="y")
ax.legend()
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
x=rng.normal(size=9000); y=0.6*x+rng.normal(scale=0.85,size=len(x))
fig,ax=plt.subplots(figsize=(8.2,4.3))
hb=ax.hexbin(x,y,gridsize=42,cmap=cc.HERI_CMAP,mincnt=1)
fig.colorbar(hb,ax=ax,label="Count")
ax.set(title="Hexbin density",xlabel="x",ylabel="y")
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
n=55
x=rng.uniform(0,10,n); y=0.5*x+rng.normal(0,1.1,n)
size=rng.uniform(30,320,n); value=rng.uniform(0,1,n)
fig,ax=plt.subplots(figsize=(8.2,4.3))
sc=ax.scatter(x,y,s=size,c=value,cmap=cc.HERI_CMAP,alpha=0.72,edgecolor="white",linewidth=0.5)
fig.colorbar(sc,ax=ax,label="Score")
ax.set(title="Bubble plot",xlabel="x",ylabel="y")
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```
