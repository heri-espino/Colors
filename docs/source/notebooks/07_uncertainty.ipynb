# Uncertainty and intervals

Error bars, confidence bands, fan charts, and interval summaries.

```python
import numpy as np
import matplotlib.pyplot as plt
import contrastcolors as cc

cc.set_style("heri", font="DejaVu Sans")
x=np.linspace(0,8,220)
mean=0.25*np.sin(x)+0.07*x
sd=0.12+0.035*np.cos(x*0.8)

fig,ax=plt.subplots(figsize=(8.2,4.2))
ax.plot(x,mean,color=cc.HERI_PALETTE[1],label="Mean")
ax.fill_between(x,mean-1.96*sd,mean+1.96*sd,color=cc.HERI_PALETTE[1],alpha=0.18,label="95% band")
ax.fill_between(x,mean-sd,mean+sd,color=cc.HERI_PALETTE[1],alpha=0.26,label="68% band")
ax.set(title="Nested uncertainty bands",xlabel="x",ylabel="Estimate")
ax.legend()
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
points=np.linspace(0,8,12)
truth=0.25*np.sin(points)+0.07*points
err=0.09+0.03*np.cos(points)
fig,ax=plt.subplots(figsize=(8.2,4.2))
ax.errorbar(points,truth,yerr=err,fmt="o",capsize=3,color=cc.HERI_PALETTE[0],ecolor=cc.HERI_NEUTRAL["mid"])
ax.set(title="Error bars",xlabel="x",ylabel="Estimate")
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
quantiles=[0.50,0.80,0.95]
alphas=[0.30,0.20,0.12]
widths=[0.08,0.15,0.24]
fig,ax=plt.subplots(figsize=(8.2,4.2))
for q,alpha,width in reversed(list(zip(quantiles,alphas,widths))):
    ax.fill_between(x,mean-width,mean+width,color=cc.HERI_PALETTE[2],alpha=alpha,label=f"{int(q*100)}%")
ax.plot(x,mean,color=cc.HERI_PALETTE[2],linewidth=2)
ax.set(title="Fan chart",xlabel="x",ylabel="Forecast")
ax.legend()
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```

```python
labels=["Model A","Model B","Model C","Model D"]
estimate=np.array([0.18,-0.06,0.31,0.11])
se=np.array([0.07,0.05,0.09,0.06])
fig,ax=plt.subplots(figsize=(7.8,3.8))
ypos=np.arange(len(labels))
ax.errorbar(estimate,ypos,xerr=1.96*se,fmt="o",capsize=3,color=cc.HERI_PALETTE[1],ecolor=cc.HERI_NEUTRAL["mid"])
ax.axvline(0,color=cc.HERI_NEUTRAL["light"],linewidth=1)
ax.set_yticks(ypos,labels)
ax.set(title="Coefficient / interval plot",xlabel="Estimate",ylabel="")
fig.tight_layout()
```

```python
cc.show_accessibility_panel(fig, max_width=640)
plt.show()
```
